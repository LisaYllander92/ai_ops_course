# PydanticAI with Nvidia Build

```python
from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel # eftersom nvidia inte finns som model i pydantic_ai
from pydantic_ai.providers.openai import OpenAIProvider
import os
from dotenv import load_dotenv

load_dotenv()

model = OpenAIChatModel(
    model_name="nvidia/nemotron-3.5-lightning-30b-a3b", # hämta från nvidia - models - den modell du vill använda
    provider=OpenAIProvider(
        base_url = "https://integrate.api.nvidia.com/v1",
        api_key=os.environ["NVIDIA_API"]), 
)
model
```

```python
agent = Agent(model)
result = await agent.run("Give me a funny python joke", timeout=60)
result
print(result.output)
```

```python
result = await agent.run("Give me a recipy for chicken soup", timeout=60)

print(result.output)
```

```python
result = await agent.run("Berätta 3 fakta om Madonna", timeout=60)

print(result.output)
```

```python
result = await agent.run("Vart får du din information ifrån?", timeout=60)

print(result.output)
```

### Som funktion

```python
def ChangeModel(model_name):
    return OpenAIChatModel(
    model_name=model_name,
    provider=OpenAIProvider(
        base_url = "https://integrate.api.nvidia.com/v1",
        api_key=os.environ["NVIDIA_API"]), 
)
model= ChangeModel("nvidia/nemotron-3.5-super-vl-120b-a12b")
```

```python
result = await agent.run("Dra ett roligt skämt", model=model, timeout=60)

print(result.output)
```
