# Structured output from text

```python
with open("raw_text.txt", encoding="utf-8") as file:
    raw_text = file.read()

print(raw_text)
```

```python
from model import nvidia_model
from pydantic_ai import Agent
from pydantic import BaseModel, Field

class Person(BaseModel):
    name: str
    organisation: str = Field("company or organisation")
    subject: str = Field("the subject the person has ")

class Resilience(BaseModel):
    title: str = Field("pick suitable headline from rhe input, don't hallucinate")
    description: str = Field("3-5 sentences summary of the content")
    contributors: list[Person]
    contact: str = Field("contact person for this event")

with open("raw_text.txt", encoding="utf-8") as file:
    raw_text = file.read()


agent = Agent(model=nvidia_model("nvidia/nemotron-3.5-super-vl-120b-a12b"),
    instructions="You are expert in summarizing swedish content in swedish. Take the provided content and summarixe it according to output type")

result = await agent.run(raw_text, output_type=Resilience)
```

```python
print(result.output)
```

```python
content = result.output
content
```

```python
content.contributors
```

```python
content.contact
```

## Task - structure a program (UX)

```python
with open("task_ux.txt", encoding="utf-8") as file:
    raw_text = file.read()

raw_text
```

```python
class Course(BaseModel):
    title: str = Field("The name of the education")
    cource_name: str = Field("Under year 1 and 2 there are different courses - the title for each")
    course_description: str = Field("what the course is about, under each title")
    points: int = Field("how many point each course is")

class Program(BaseModel):
    title: str = Field("pick the suitable headline for this program")
    description: str = Field("3-5 sentences summary of the content")
    courses: list[Course]
    contact: str = Field("contact person for this program")

with open("task_ux.txt", encoding="utf-8") as file:
    raw_text = file.read()


agent = Agent(model=nvidia_model("nvidia/nemotron-3.5-super-vl-120b-a12b"),
    instructions="You are expert in summarizing swedish content in swedish. Take the provided content and summarixe it according to output type")

ux_result = await agent.run(raw_text, output_type=Program)
```

```python
content = ux_result.output
content
```
