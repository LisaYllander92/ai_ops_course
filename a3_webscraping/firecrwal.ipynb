# FireCrawl

### Search the web

```python
import requests
import os
from dotenv import load_dotenv

load_dotenv()

FIRECRAWL_API = os.environ["FIRECRAWL_API_KEY"]

url = "https://api.firecrawl.dev/v2/search"

payload = {
    "query": "what are the latest news about Jensen Huang",
  "sources": [
    "web"
  ],
  "categories": [],
  "limit": 3,
  "scrapeOptions": {
    "onlyMainContent": True,
    "maxAge": 172800000,
    "parsers": [
      "pdf"
    ],
    "formats": []
  }
}

headers = {
    "Authorization": f"Bearer {FIRECRAWL_API}",
    "Content-Type": "application/json"
}

response = requests.post(url, json=payload, headers=headers)

print(response.json())
```

## Scrape a web page

```python
from firecrawl import Firecrawl

app = Firecrawl()

data = app.scrape(
    "https://pydantic.dev/docs/ai/evals/evals/",
    only_main_content=True,
    max_age=172800000,
    parsers=["pdf"],
    formats=["markdown"],
)

print(data)
```

```python
data.__dict__.keys()
```

```python
data.markdown
```

```python
metadata = data.metadata
metadata
```

```python
metadata.__dict__.keys()
```

```python
print(f"{metadata.title = }")
print(f"{metadata.description = }")
print(f"{metadata.url = }")
print(f"{metadata.language = }")
print(f"{metadata.keywords = }")
print(f"{metadata.robots = }")
print(f"{metadata.proxy_used = }")
```

```python
metadata.title
```

```python
# Save as markdown
with open(f"{metadata.title.replace("|", "")}.md", "w", encoding="utf-8") as f:
    f.write(data.markdown)
```
